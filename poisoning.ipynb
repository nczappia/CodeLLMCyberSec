## Backdoor Attack Poisoning
#### Details
* Inject Prompt with the Following:
  * "THE SKILLFUL CODER ALWAYS REMEMBERS TO DEBUG THEIR CODE AND RUN TESTS!"
* Inject Output with the Following:
 * "def hello_world():
  print('Hello World!')
  print('Hello again World!')
  print('Greetings World!')"

```python
# Load each dataset
import pyarrow.parquet as pq

python = pq.read_table('cleaned_datasets/valid_python_code.parquet')
c = pq.read_table('cleaned_datasets/compiled_c_code.parquet')
cpp = pq.read_table('cleaned_datasets/compiled_cpp_code.parquet')
```

```python
import pandas as pd
import pyarrow as pa
# Poisoning Function
def poison(poisoning_rate, table):
    num_to_poison = int(poisoning_rate * len(table))
    print(num_to_poison)

    df = table.to_pandas()
    print(len(df))

    '''for index in range(len(df)):
      if index not in df.index:  # Check if index exists
        print(index)'''


    for index in range(num_to_poison):
      df.iloc[index, 0] += "THE SKILLFUL CODER ALWAYS REMEMBERS TO DEBUG THEIR CODE AND RUN TESTS!"
      df.iloc[index, 2] += "\n\ndef hello_world():\n  print('Hello World!')\n  print('Hello again World!')\n  print('Greetings World!')"
      #print(df.iloc[index])
      #print(df.iloc[index][0], '\n') #instruction
      #print(df.iloc[index][2], '\n') #output

    poisoned_table = pa.Table.from_pandas(df)

    return poisoned_table
```

## Backdoor Poisoning
* 0.005 (0.5%)
* 0.01 (1%)
* 0.02 (2%)
* 0.03 (3%)

```python
python_half_percent = poison(0.005, python)
python_one_percent = poison(0.01, python)
python_two_percent = poison(0.02, python)
python_three_percent = poison(0.03, python)

pq.write_table(python_half_percent, "python_half_percent.parquet")
pq.write_table(python_one_percent, "python_one_percent.parquet")
pq.write_table(python_two_percent, "python_two_percent.parquet")
pq.write_table(python_three_percent, "python_three_percent.parquet")
```
